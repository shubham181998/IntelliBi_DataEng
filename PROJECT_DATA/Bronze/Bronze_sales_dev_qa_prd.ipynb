```python
TableName = f"saleslake_{env}.bronze_{env}.rawsales"
print(TableName)
srcFileLoc=f"/Volumes/saleslake_{env}/bronze_{env}/vol_saleslake_src_files_{env}/daily_sales/"
print(srcFileLoc)

```

```python
spark.sql(
f"""COPY INTO {TableName}
    FROM 
    (
        SELECT sale_id, product, category, quantity, price, sale_date, region, 
        current_timestamp() AS ingest_ts FROM "{srcFileLoc}"
    )
    FILEFORMAT = CSV
    FORMAT_OPTIONS ('header' = 'true')
    COPY_OPTIONS ('mergeSchema' = 'false')
"""
)
```
