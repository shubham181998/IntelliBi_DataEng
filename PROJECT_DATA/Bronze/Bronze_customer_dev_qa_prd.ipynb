```sql
%python
TableName= f"saleslake_{Env}.bronze_{Env}.rawcustomer"
print(TableName)
srcFileLoc = f"/Volumes/saleslake_{Env}/bronze_{Env}/vol_saleslake_src_files_{Env}/daily_customer"
print(srcFileLoc)
```

```sql
%python
spark.sql (f"""
COPY INTO {TableName}
FROM (SELECT customer_id, customer_name, email,  phone,  address, city, state, country, zip_code, segment, current_timestamp() AS ingest_ts
FROM '{srcFileLoc}')
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'false')  
""")
```
